```python
from pathlib import Path
import calendar

import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

NOTEBOOK_DIR = Path.cwd()
DATA_DIR = NOTEBOOK_DIR / 'SI_results'
OUTPUT_DIR = NOTEBOOK_DIR / 'SI_Figures'

DATA_FILES = {
    'Harbin': DATA_DIR / 'FigS1_harbin.csv',
    'Shanghai': DATA_DIR / 'FigS1_shanghai.csv',
}

for city, path in DATA_FILES.items():
    if not path.exists():
        raise FileNotFoundError(f'Cannot find {city} S1 data: {path}')

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
```

```python
# ===============================
# Global journal-style settings
# ===============================
plt.rcParams.update({
    'font.family': 'Arial',
    'font.size': 14,
    'axes.labelsize': 14,
    'axes.titlesize': 14,
    'xtick.labelsize': 12,
    'ytick.labelsize': 12,
    'legend.fontsize': 12,
    'figure.facecolor': 'white',
    'axes.facecolor': 'white',
    'axes.linewidth': 2,
    'xtick.direction': 'out',
    'ytick.direction': 'out',
    'axes.spines.top': False,
    'axes.spines.right': False,
    'xtick.major.size': 6,
    'ytick.major.size': 6,
    'xtick.major.width': 2.0,
    'ytick.major.width': 2.0,
    'xtick.minor.visible': False,
    'ytick.minor.visible': False,
    'lines.solid_capstyle': 'round',
    'axes.unicode_minus': False,
})

BLUE_CMAP = mpl.colors.LinearSegmentedColormap.from_list(
    'white_to_framework_blue',
    ['#FFFFFF', '#DCEBF3', '#7FB3D5', '#2171A0']
)
RED_CMAP = mpl.colors.LinearSegmentedColormap.from_list(
    'white_to_framework_red',
    ['#FFFFFF', '#F6D6D8', '#EE8A8F', '#DA4048']
)

PANEL_STYLE = {
    'cmap': {'Harbin': BLUE_CMAP, 'Shanghai': RED_CMAP},
    'bar_edge': '#4A4A4A',
    'grid': '#D4D4D4',
    'spine': '#1F1F1F',
    'label': '#1F1F1F',
}

PANEL_CONFIG = {
    ('Harbin', 'total_orders'): {
        'label': 'a',
        'norm_range': (1000, 4000),
        'cbar_label': 'Daily average order volumes',
    },
    ('Shanghai', 'total_orders'): {
        'label': 'b',
        'norm_range': (3000, 7000),
        'cbar_label': 'Daily average order volumes',
    },
    ('Harbin', 'attendance'): {
        'label': 'c',
        'norm_range': (20, 100),
        'cbar_label': 'Daily average attendance (riders)',
    },
    ('Shanghai', 'attendance'): {
        'label': 'd',
        'norm_range': (80, 140),
        'cbar_label': 'Daily average attendance (riders)',
    },
    ('Harbin', 'orders_per_rider'): {
        'label': 'e',
        'norm_range': (30, 50),
        'cbar_label': 'Daily order volume per rider',
    },
    ('Shanghai', 'orders_per_rider'): {
        'label': 'f',
        'norm_range': (30, 60),
        'cbar_label': 'Daily order volume per rider',
    },
}
```

```python
def load_s1_daily_data(path):
    df = pd.read_csv(path, parse_dates=['date'])
    required_cols = {'date', 'total_orders', 'attendance', 'orders_per_rider'}
    missing_cols = required_cols.difference(df.columns)
    if missing_cols:
        raise ValueError(f'{path.name} is missing columns: {sorted(missing_cols)}')

    df = df.copy()
    df['month'] = df['date'].dt.month
    df['weekday'] = df['date'].dt.weekday + 1
    return df.reset_index(drop=True)


daily_data = {city: load_s1_daily_data(path) for city, path in DATA_FILES.items()}
{city: df.shape for city, df in daily_data.items()}
```

```python
def aggregate_month_weekday(df, value_col):
    full_index = pd.MultiIndex.from_product(
        [range(1, 13), range(1, 8)], names=['month', 'weekday']
    )
    stats = (
        df.groupby(['month', 'weekday'])[value_col]
        .mean()
        .reindex(full_index)
        .reset_index()
    )
    return stats


def add_polar_guides(ax, max_radius):
    theta_month = 2 * np.pi / 12
    for month in range(12):
        theta = month * theta_month
        ax.plot([theta, theta], [0, max_radius], color=PANEL_STYLE['grid'], lw=0.9, zorder=0)

    grid_values = np.linspace(0, max_radius, 5)[1:]
    ax.set_yticks(grid_values)
    ax.set_yticklabels([])
    ax.yaxis.grid(True, color=PANEL_STYLE['grid'], lw=0.9)
    ax.xaxis.grid(False)


def plot_overall_polar(ax, df, city, value_col):
    config = PANEL_CONFIG[(city, value_col)]
    stats = aggregate_month_weekday(df, value_col)
    theta_month = 2 * np.pi / 12
    width = theta_month / 7 * 0.86
    norm = mpl.colors.Normalize(vmin=config['norm_range'][0], vmax=config['norm_range'][1])
    cmap = PANEL_STYLE['cmap'][city]
    max_radius = config['norm_range'][1] * 1.04

    ax.set_theta_direction(-1)
    ax.set_theta_offset(np.pi / 2)
    ax.set_ylim(0, max_radius)
    add_polar_guides(ax, max_radius)

    for _, row in stats.dropna(subset=[value_col]).iterrows():
        theta = (row['month'] - 1) * theta_month + (row['weekday'] - 0.5) * (theta_month / 7)
        radius = row[value_col]
        ax.bar(
            theta,
            radius,
            width=width,
            color=cmap(norm(radius)),
            edgecolor=PANEL_STYLE['bar_edge'],
            linewidth=0.45,
            alpha=0.9,
            zorder=3,
        )

    ax.set_xticks([i * theta_month for i in range(12)])
    ax.set_xticklabels(calendar.month_abbr[1:], color=PANEL_STYLE['label'], fontname='Arial')
    ax.tick_params(axis='x', pad=8, width=2.0)
    ax.spines['polar'].set_color(PANEL_STYLE['spine'])
    ax.spines['polar'].set_linewidth(2.0)

    ax.text(
        -0.08, 1.02, config['label'],
        transform=ax.transAxes,
        fontsize=20,
        fontweight='bold',
        va='top',
        ha='left',
    )
    ax.set_title(city, pad=18, fontname='Arial')

    sm = mpl.cm.ScalarMappable(cmap=cmap, norm=norm)
    sm.set_array([])
    cbar = ax.figure.colorbar(sm, ax=ax, pad=0.09, shrink=0.74)
    cbar.set_label(config['cbar_label'], rotation=90, labelpad=14, fontname='Arial')
    cbar.outline.set_linewidth(1.2)
    cbar.ax.tick_params(length=6, width=2.0)
    return ax
```

```python
cities = ['Harbin', 'Shanghai']
metrics = ['total_orders', 'attendance', 'orders_per_rider']

fig, axes = plt.subplots(
    nrows=3,
    ncols=2,
    figsize=(15, 21),
    dpi=600,
    subplot_kw={'projection': 'polar'},
)

for row, value_col in enumerate(metrics):
    for col, city in enumerate(cities):
        plot_overall_polar(axes[row, col], daily_data[city], city, value_col)

plt.tight_layout()
output_path = OUTPUT_DIR / 'FigS1.tiff'
fig.savefig(output_path, format='tiff', dpi=600, bbox_inches='tight')
plt.show()
output_path
```
